```python
marks=[]
for i in range (1,6):
    while True:
        m=float(input(f"Enter marks for subject {i} : "))
        if 0<=m<=100:
            marks.append(m)
            break
        else:
            print("Invalid input.please input valid marks between 0 and 100")
total=sum(marks)
average=total/5
if average>=90:
    grade="A+"
elif average>=80:
    grade="A"
elif average>=70:
    grade="B"
elif average>=60:
    grade="C"
elif average>=50:
    grade="D"
else:
    grade="Fail"
print("\nTotal Marks: ",total)
print("Average Marks:",average)
print("Grade:",grade)

```
